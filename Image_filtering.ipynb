```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image
img = cv2.imread('/content/image1.jpg', cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(6, 5))
plt.imshow(img, cmap='gray')
plt.title("Original Input Image")
plt.axis('off')
```

```python
#sobar filter
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sobel_x, sobel_y)
sobel = cv2.convertScaleAbs(sobel)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(sobel, cmap='gray')
plt.title("Sobel Filter")
plt.axis('off')
```

```python
#Box Filter
box = cv2.boxFilter(img, -1, (3, 3))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(box, cmap='gray')
plt.title("Box Filter")
plt.axis('off')
```

```python
#weighted average filter
kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32) / 16

weighted = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(weighted, cmap='gray')
plt.title("Weighted Average Filter")
plt.axis('off')
```

```python
#Gaussian Filter
gaussian = cv2.GaussianBlur(img, (5, 5), 0)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(gaussian, cmap='gray')
plt.title("Gaussian Filter")
plt.axis('off')
```

```python
#sharpen_kernel
sharpen_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

sharpen = cv2.filter2D(img, -1, sharpen_kernel)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sharpen, cmap='gray')
plt.title("Sharpening Filter")
plt.axis('off')
plt.show()
```

```python
#min_filter
min_filter = cv2.erode(img, np.ones((3, 3), np.uint8))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(min_filter, cmap='gray')
plt.title("Min Filter")
plt.axis('off')
plt.show()
```

```python
#Max Filter
max_filter = cv2.dilate(img, np.ones((3, 3), np.uint8))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(max_filter, cmap='gray')
plt.title("Max Filter")
plt.axis('off')
```

```python
#mean_filter
mean_filter = cv2.blur(img, (3, 3))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(mean_filter, cmap='gray')
plt.title("Mean Filter")
plt.axis('off')
plt.show()
```

```python
#median_filter
median_filter = cv2.medianBlur(img, 3)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Input Image")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(median_filter, cmap='gray')
plt.title("Median Filter")
plt.axis('off')
plt.show()
```

```python
filters = [
    ("Original Image", img),
    ("Sobel Filter", sobel),
    ("Box Filter", box),
    ("Weighted Average", weighted),
    ("Gaussian Filter", gaussian),
    ("Sharpening Filter", sharpen),
    ("Min Filter", min_filter),
    ("Max Filter", max_filter),
    ("Mean Filter", mean_filter),
    ("Median Filter", median_filter)
]

plt.figure(figsize=(13, 12))

for i, (title, result) in enumerate(filters):
    plt.subplot(2, 5, i + 1)
    plt.imshow(result, cmap='gray')
    plt.title(title)
    plt.axis('off')

plt.suptitle("Task 4: Image Filtering Results", fontsize=16)
plt.tight_layout()
plt.show()
```
