```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Give image paths
img1_path = '/content/image1.jpg'
img2_path = '/content/black_and_white.jfif'
# Read images in grayscale
img1 = cv2.imread(img1_path, cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(img2_path, cv2.IMREAD_GRAYSCALE)
# Resize both images to same size
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

print("Two images loaded successfully!")
```

```python
addition = cv2.add(img1, img2)
```

```python
subtraction = cv2.subtract(img1, img2)
```

```python
multiplication = cv2.multiply(img1, img2, scale=1/255)
```

```python
division = cv2.divide(img1, img2)
```

```python
plt.figure(figsize=(12, 10))
# Original Image 1
plt.subplot(2, 3, 1)
plt.imshow(img1, cmap='gray')
plt.title("Original Image 1")
plt.axis('off')
# Original Image 2
plt.subplot(2, 3, 2)
plt.imshow(img2, cmap='gray')
plt.title("Original Image 2")
plt.axis('off')
# Addition
plt.subplot(2, 3, 3)
plt.imshow(addition, cmap='gray')
plt.title("Addition")
plt.axis('off')
# Subtraction
plt.subplot(2, 3, 4)
plt.imshow(subtraction, cmap='gray')
plt.title("Subtraction")
plt.axis('off')
# Multiplication
plt.subplot(2, 3, 5)
plt.imshow(multiplication, cmap='gray')
plt.title("Multiplication")
plt.axis('off')
# Division
plt.subplot(2, 3, 6)
plt.imshow(division, cmap='gray')
plt.title("Division")
plt.axis('off')
plt.suptitle("Task 3: Arithmetic Operations", fontsize=16)
plt.tight_layout()
plt.show()
```
