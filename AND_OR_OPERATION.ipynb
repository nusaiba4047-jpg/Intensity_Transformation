```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Image path
img_path = '/content/black_and_white.jfif'

# Read image in grayscale
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.axis("off")
plt.show()
```

```python

# Convert image into binary image
_, binary1 = cv2.threshold(
    img, 127, 255, cv2.THRESH_BINARY
)

# Display binary image
plt.figure(figsize=(5, 4))
plt.imshow(binary1, cmap='gray')
plt.title("Binary Image A")
plt.axis("off")
plt.show()
```

```python

# Create second image by flipping horizontally
img2 = cv2.flip(img, 1)

# Convert second image into binary
_, binary2 = cv2.threshold(
    img2, 127, 255, cv2.THRESH_BINARY
)

# Display both binary images
plt.figure(figsize=(8, 5))

plt.subplot(1, 2, 1)
plt.imshow(binary1, cmap='gray')
plt.title("Binary Image A")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary2, cmap='gray')
plt.title("Binary Image B")
plt.axis("off")

plt.show()
```

```python

# AND operation
and_result = cv2.bitwise_and(binary1, binary2)

# OR operation
or_result = cv2.bitwise_or(binary1, binary2)

# NOT operation
not_result = cv2.bitwise_not(binary1)

# Display all results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(binary1, cmap='gray')
plt.title("Original Binary Image A")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(and_result, cmap='gray')
plt.title("AND Operation")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(or_result, cmap='gray')
plt.title("OR Operation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(not_result, cmap='gray')
plt.title("NOT Operation")
plt.axis("off")

plt.tight_layout()
plt.savefig("Task2_Logical_Operations.png", dpi=150)
plt.show()
```

```python

cv2.imwrite("AND_Result.png", and_result)
cv2.imwrite("OR_Result.png", or_result)
cv2.imwrite("NOT_Result.png", not_result)

print("All logical operations completed successfully!")

from google.colab import files

files.download("Task2_Logical_Operations.png")
```
