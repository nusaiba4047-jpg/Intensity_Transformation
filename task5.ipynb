```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image
img = cv2.imread('/content/image1.jpg', cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(6, 5))
plt.imshow(img, cmap='gray')
plt.title("Original Input Image")
plt.axis('off')
```

```python
# Apply Laplacian filter
laplacian = cv2.Laplacian(img, cv2.CV_64F)

# Convert to displayable image
laplacian_output = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(laplacian_output, cmap='gray')
plt.title("Laplacian Image")
plt.axis('off')

plt.show()
```

```python
# Enhanced Laplacian

# Convert original image to float64
img_float = img.astype(np.float64)

# Apply Enhanced Laplacian
enhanced_laplacian = img_float - laplacian

# Convert result to uint8 (0-255)
enhanced_laplacian = np.clip(
    enhanced_laplacian, 0, 255
).astype(np.uint8)

# Display output
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(enhanced_laplacian, cmap='gray')
plt.title("Enhanced Laplacian")
plt.axis('off')

plt.show()
```

```python
# Create blurred image
blurred = cv2.GaussianBlur(img, (5, 5), 0)

# High-boost mask
A = 1.5

high_boost = cv2.addWeighted(
    img, A,
    blurred, -(A - 1),
    0
)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(high_boost, cmap='gray')
plt.title("High-Boost Filter")
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 4, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(laplacian_output, cmap='gray')
plt.title("Laplacian")
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(enhanced_laplacian, cmap='gray')
plt.title("Enhanced Laplacian")
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(high_boost, cmap='gray')
plt.title("High-Boost")
plt.axis('off')

plt.suptitle("Task 5: Image Enhancement")
plt.tight_layout()
plt.show()
```
